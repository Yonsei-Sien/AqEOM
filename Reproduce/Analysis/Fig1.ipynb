```python
import matplotlib.colors as mcolors
import matplotlib.ticker as ticker
import matplotlib.pyplot as plt
import numpy as np

def RMSD(y, y_ref):
    N_var   = y_ref.shape[0] * y_ref.shape[1]
    RMSE    = np.sqrt(np.sum((y-y_ref)**2) / N_var)
    return np.array(RMSE)
```

## Energy

```python
fig, axs = plt.subplots(1, 2, figsize=(25, 8), dpi=1000)
plt.subplots_adjust(wspace=0.3)
plt.rc('font', size=40)

master_dir  = "***/Data"
colors      = ['#003876', '#00763e', '#760038', '#763e00']
dist_lst    = np.around(np.arange(0.5, 2.01, 0.1), 3)

# Data Load
energy_cas      = np.zeros((len(dist_lst), 3))
energy_QSE      = np.zeros((len(dist_lst), 3))
energy_qEOM     = np.zeros((len(dist_lst), 3))
energy_fqEOM    = np.zeros((len(dist_lst), 3))
for inddist, dist in enumerate(dist_lst):
    energy_cas[inddist]     = np.load(f"{master_dir}/H2/Energy/{dist}_CAS.npy")[:3]
    energy_QSE[inddist]     = np.load(f"{master_dir}/H2_QSE/Energy/{dist}_QSE.npy")
    energy_qEOM[inddist]    = np.load(f"{master_dir}/H2/Energy/{dist}_CASCI_MATCH_qEOM.npy")
    energy_fqEOM[inddist]   = np.load(f"{master_dir}/H2/Energy/{dist}_CASCI_MATCH_fqEOM.npy")

print('QSE', RMSD(energy_QSE, energy_cas))
print('qEOM', RMSD(energy_qEOM, energy_cas))
print('AqEOM', RMSD(energy_fqEOM, energy_cas))

ax = axs[0]
for ex in range(3):
    # Reference
    ax.plot(dist_lst, energy_cas[:, ex]  , color=colors[ex], linewidth=5, markeredgecolor='black')
    ax.plot(dist_lst, energy_fqEOM[:, ex], marker='o', markersize=15, linewidth=0, markeredgecolor=colors[ex], markerfacecolor='none', markeredgewidth=3)
    ax.plot(dist_lst, energy_qEOM[:, ex] , color=colors[ex], marker='x', markersize=16, linewidth=0, markeredgecolor=colors[ex], mew=3.5)
    ax.plot(dist_lst, energy_QSE[:, ex]  , marker='d', markersize=15, linewidth=0, markeredgecolor=colors[ex], markerfacecolor='none', markeredgewidth=3)
    
ax.set_xlabel('Distance ($\AA$)')
ax.grid()
ax.set_xticks([0.5, 1.0, 1.5, 2.0]) 
ax.set_yticks([-1.0, -0.5, 0, 0.5, 1.0]) 
ax.set_ylabel('$E_n^{O/U}$\n(Hartree)')

ax = axs[1]
for ex in range(3):
    # Reference
    ax.plot(dist_lst, abs(energy_fqEOM[:, ex] - energy_cas[:, ex]), marker='o', markersize=15, linewidth=0, markeredgecolor=colors[ex], markerfacecolor='none', markeredgewidth=3)
    ax.plot(dist_lst, abs(energy_qEOM[:, ex]  - energy_cas[:, ex]), color=colors[ex], marker='x', markersize=16, linewidth=0, markeredgecolor=colors[ex], mew=3.5)
    ax.plot(dist_lst, abs(energy_QSE[:, ex]   - energy_cas[:, ex]), marker='d', markersize=15, linewidth=0, markeredgecolor=colors[ex], markerfacecolor='none', markeredgewidth=3)

ax.grid()
ax.set_xlabel('Distance ($\AA$)')
ax.set_xticks([0.5, 1.0, 1.5, 2.0])
ax.set_ylabel('$E_n^{O/U}$ Error\n(Hartree)')

plt.savefig(f'{master_dir}/Figs/Fig1_E.png', bbox_inches='tight', dpi=1000)
```

## Dipole moment

```python
plt.figure(figsize=(25, 8), dpi=1000)
plt.subplots_adjust(wspace=0.3)
plt.rc('font', size=40)

master_dir  = "***/Data"
colors      = ['#003876', '#00763e', '#760038', '#763e00']
dist_lst    = np.around(np.arange(0.5, 2.01, 0.1), 3)

# Data Load
dipole_cas          = np.zeros((len(dist_lst), 3, 3, 3))
dipole_QSE          = np.zeros((len(dist_lst), 3, 3, 3))
dipole_qEOM         = np.zeros((len(dist_lst), 3, 3, 3))
dipole_fqEOM        = np.zeros((len(dist_lst), 3, 3, 3))
for inddist, dist in enumerate(dist_lst):
    dipole_cas[inddist]     = abs(np.load(f"{master_dir}/H2/Dipole/{dist}_CAS.npy")) / 0.393430
    dipole_QSE[inddist]     = abs(np.load(f"{master_dir}/H2_QSE/Dipole/{dist}_QSE.npy")) / 0.393430
    dipole_qEOM[inddist]    = abs(np.load(f"{master_dir}/H2/Dipole/{dist}_CASCI_MATCH_qEOM.npy")) / 0.393430
    dipole_fqEOM[inddist]   = abs(np.load(f"{master_dir}/H2/Dipole/{dist}_CASCI_MATCH_fqEOM.npy")) / 0.393430

dip_cas             = np.zeros((len(dist_lst), sum(range(4))))
dip_QSE             = np.zeros((len(dist_lst), sum(range(4))))
dip_qEOM            = np.zeros((len(dist_lst), sum(range(4))))
dip_fqEOM           = np.zeros((len(dist_lst), sum(range(4))))
master_count        = 0
for Rind in range(3):
    for Lind in range(Rind + 1):
        dip_cas[:, master_count]    = dipole_cas[:, Lind, Rind, 2]  
        dip_QSE[:, master_count]    = dipole_QSE[:, Lind, Rind, 2]
        dip_qEOM[:, master_count]   = dipole_qEOM[:, Lind, Rind, 2] 
        dip_fqEOM[:, master_count]  = dipole_fqEOM[:, Lind, Rind, 2] 
        master_count               += 1
print('QSE', RMSD(dip_QSE, dip_cas))
print('qEOM', RMSD(dip_qEOM, dip_cas))
print('AqEOM', RMSD(dip_fqEOM, dip_cas))

# Plot
for i in range(1, 3):
    plt.subplot(120+i)
    for ex in range(i + 1):
        # Reference
        plt.plot(dist_lst, dipole_cas[:, ex, i, 2]  , color=colors[ex], linewidth=5, markeredgecolor='black')
        plt.plot(dist_lst, dipole_fqEOM[:, ex, i, 2], marker='o', markersize=15, linewidth=0, markeredgecolor=colors[ex], markerfacecolor='none', markeredgewidth=3)
        plt.plot(dist_lst, dipole_qEOM[:, ex, i, 2] , color=colors[ex], marker='x', markersize=16, linewidth=0, markeredgecolor=colors[ex], mew=3.5)
        plt.plot(dist_lst, dipole_QSE[:, ex, i, 2]  , marker='d', markersize=15, linewidth=0, markeredgecolor=colors[ex], markerfacecolor='none', markeredgewidth=3)
    plt.xlabel('Distance ($\AA$)')
    plt.grid()
    plt.xticks([0.5, 1.0, 1.5, 2.0]) 
    plt.yticks([0, 5, 10]) 

    if i == 1:
        plt.ylabel('$|\mu_Z|$\n(Debye)')
        plt.plot([], [], linewidth=5, color='black', label='CASCI')
        plt.plot([], [], linewidth=0, color='black', marker='x', markersize=16, markeredgecolor='black', mew=3.5, label='qEOM')
        plt.plot([], [], linewidth=0, markersize=15, marker='o', markeredgecolor='black', markerfacecolor='none', markeredgewidth=3, label='AqEOM')
        plt.plot([], [], linewidth=0, markersize=15, marker='d', markeredgecolor='black', markerfacecolor='none', markeredgewidth=3, label='QSE')
        plt.legend(ncol=2, frameon=False, bbox_to_anchor=(1.6, -0.3))

plt.savefig(f'{master_dir}/Figs/Fig1_D.png', bbox_inches='tight', dpi=1000)
```
