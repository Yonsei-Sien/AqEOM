```python
import matplotlib.colors as mcolors
import matplotlib.ticker as ticker
import matplotlib.pyplot as plt
import numpy as np

def Analytics(y, y_ref):
    N_var   = y_ref.shape[0] * y_ref.shape[1]
    MAE     = np.sum(abs(y-y_ref)) / N_var
    RMSE    = np.sqrt(np.sum((y-y_ref)**2) / N_var)
    return np.array([MAE, RMSE])

def rsquare(exp, fit):
    mean    = np.sum(exp) / len(exp)
    SST     = np.sum((exp - mean)**2)
    SSR     = np.sum((exp - fit)**2)
    return 1 - SSR/SST
```

## DC-r2SCAN

```python
fig, subplots = plt.subplots(2, 1, figsize=(20, 8), height_ratios=[2, 5], dpi=1000)
plt.subplots_adjust(hspace=0, wspace=0.2)
plt.rc('font', size=32)

master_dir  = "***/Data"
colors      = ['#003876', '#00763e', '#760038', '#763e00']
dist_lst    = np.around(np.arange(0.5, 2.01, 0.1), 3)

# Data Load
energy_cas          = np.zeros((len(dist_lst), 3))
energy_qEOM         = np.zeros((len(dist_lst), 3))
energy_fqEOM        = np.zeros((len(dist_lst), 3))
for inddist, dist in enumerate(dist_lst):
    energy_cas[inddist]         = np.load(f"{master_dir}/H2/Energy/{dist}_CAS_r2SCAN.npy")
    energy_qEOM[inddist]        = np.load(f"{master_dir}/H2/Energy/{dist}_CASCI_MATCH_qEOM_r2SCAN.npy")
    energy_fqEOM[inddist]       = np.load(f"{master_dir}/H2/Energy/{dist}_CASCI_MATCH_fqEOM_r2SCAN.npy")

print(Analytics(energy_qEOM , energy_cas) * 627.509)
print(Analytics(energy_fqEOM, energy_cas) * 627.509)

e_sub   = subplots[1]
err_sub = subplots[0]
for ex in range(3):
    E_ref   = energy_cas[:, ex]
    # Reference
    e_sub.plot(  dist_lst, E_ref                         , color=colors[ex], linewidth=5, markeredgecolor='black')

    # AqEOM
    e_sub.plot(  dist_lst, energy_fqEOM[:, ex]           , marker='o', markersize=15, linewidth=0, markeredgecolor=colors[ex], markerfacecolor='none', markeredgewidth=3)
    err_sub.plot(dist_lst, abs(energy_fqEOM[:, ex]-E_ref), marker='o', markersize=15, linewidth=0, markeredgecolor=colors[ex], markerfacecolor='none', markeredgewidth=3)
        
    # qEOM
    e_sub.plot(  dist_lst, energy_qEOM[:, ex]            , color=colors[ex], marker='x', markersize=16, linewidth=0, markeredgecolor=colors[ex], mew=3.5)
    err_sub.plot(dist_lst, abs(energy_qEOM[:, ex]-E_ref) , color=colors[ex], marker='x', markersize=16, linewidth=0, markeredgecolor=colors[ex], mew=3.5)
    
e_sub.set_xlabel('Distance ($\AA$)')
e_sub.grid()
err_sub.grid()
err_sub.set_yscale('log')
err_sub.set_xticks([0.5, 1.0, 1.5, 2.0], labels=[], fontsize=0, color='white')
err_sub.set_yticks([1e-12, 1e-9, 1e-6, 1e-3, 0])
e_sub.set_xticks([0.5, 1.0, 1.5, 2.0]) 
e_sub.set_yticks([-1.0, -0.5, 0, 0.5, 1.0]) 
err_sub.axhspan(0, 1/627.509, color='lightgray')

locmin = ticker.LogLocator(base=10.0, subs=np.arange(2, 10), numticks=100)
err_sub.tick_params(axis='y', which='major', length=6, color='black')
err_sub.yaxis.set_minor_locator(locmin)
err_sub.yaxis.set_minor_formatter(ticker.NullFormatter())

#err_sub.set_yticks([1e-9, 1e-7, 1e-5, 1e-3, 1e-1, 1e+1])
#err_sub.set_yticks([1e-8, 1e-6, 1e-4, 1e-2, 1], minor=True, labels=[], color='white')

# Settings
e_sub.plot([], [], linewidth=5, color='black', label='CASCI-r$^2$SCAN')
e_sub.plot([], [], linewidth=0, color='black', marker='x', markersize=16, markeredgecolor='black', mew=3.5, label='qEOM-r$^2$SCAN')
e_sub.plot([], [], linewidth=0, markersize=15, marker='o', markeredgecolor='black', markerfacecolor='none', markeredgewidth=3, label='AqEOM-r$^2$SCAN')
e_sub.legend(ncol=3, frameon=False, bbox_to_anchor=(0.96, -0.3))
e_sub.set_ylabel('Energy\n(Hartree)')
err_sub.set_ylabel('E Error\n(Hartree)')

#plt.show()
fig.align_ylabels()
plt.savefig('***/Data/Figs/Fig4_DC.png', bbox_inches='tight', dpi=1000)
```
