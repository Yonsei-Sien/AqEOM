```python
import matplotlib.colors as mcolors
import matplotlib.ticker as ticker
import matplotlib.pyplot as plt
import numpy as np
```

## vs E Err

```python
fig, axs = plt.subplots(2, 1, figsize=(10, 18), dpi=1000)
plt.subplots_adjust(hspace=0.2)
plt.rc('font', size=32)

master_dir      = "***/Data"
colors          = ['#003876', '#00763e', '#760038', '#763e00']
state_label     = ['GS', '1st', '2nd', '3rd']
markers         = ['o', '^', 's']
dist            = np.around(np.arange(0.5, 2.01, 0.1), 3)

# Load
E_CAS       = np.zeros((len(dist), 3))
D_CAS       = np.zeros((len(dist), 3, 3))
E_qEOM      = np.zeros_like(E_CAS)
D_qEOM      = np.zeros_like(D_CAS)
S_qEOM      = np.zeros_like(E_CAS)
for inddist, dist in enumerate(dist):
    E_CAS[inddist]  = np.load(f"{master_dir}/H2/Energy/{dist}_CAS.npy")[:3]
    E_qEOM[inddist] = np.load(f"{master_dir}/H2/Energy/{dist}_CASCI_MATCH_qEOM.npy")
    D_CAS[inddist]  = np.load(f"{master_dir}/H2/Dipole/{dist}_CAS.npy")[:3, :3, 2]
    D_qEOM[inddist] = np.load(f"{master_dir}/H2/Dipole/{dist}_CASCI_MATCH_qEOM.npy")[:,:,2]
    S_qEOM[inddist] = np.load(f"{master_dir}/H2/VAC/{dist}_CASCI_MATCH_qEOM.npy")[0]
E_err       = abs(E_CAS - E_qEOM)
D_err       = abs(abs(D_CAS) - abs(D_qEOM)) / 0.393430

# Energy
plt.subplot(211)
for index in range(1, 3):
    plt.plot(E_err[:, index], abs(S_qEOM[:, index]), marker=markers[index], color='none', markersize=15, linewidth=0, markeredgecolor=colors[index], markeredgewidth=3)
plt.yticks([0, 0.2, 0.4, 0.6], ['0.00', '0.20', '0.40', '0.60'])
plt.xticks([0, 0.1, 0.2, 0.3], fontsize=0, color='white')
#plt.xlabel('E Error\n(Hartree)')
plt.ylabel('$|S_{0n}|$\n(Arb. Unit)')
plt.grid()

# Dipole
plt.subplot(212)
for Rex in range(3):
    for Lex in range(Rex + 1):
        plt.plot(0.5*(E_err[:, Rex] + E_err[:, Lex]), D_err[:, Lex, Rex], markerfacecolor='none', marker=markers[Lex], markersize=15, linewidth=0, markeredgecolor=colors[Rex], markeredgewidth=3)
plt.xlabel('$E$ Error\n(Hartree)')
plt.ylabel('$|\Delta\mu_Z|$\n(Debye)')
plt.grid()

for i in range(3):
    plt.plot([], [], color=colors[i], marker='s', markersize=15, linewidth=0, label=f'$|{i}$' + r'$\rangle$')
    plt.plot([], [], markerfacecolor='none', marker=markers[i], markersize=15, linewidth=0, markeredgecolor='black', markeredgewidth=3, label=r'$\langle$' + f'${i}|$')
plt.legend(ncol=3, frameon=False, bbox_to_anchor=(0.87, -0.3))

fig.align_ylabels()
plt.savefig('***/Data/Figs/Fig2_vsE.png', bbox_inches='tight', dpi=1000)
```

## vs Hermitian

```python
fig, axs = plt.subplots(2, 1, figsize=(10, 18), dpi=1000)
plt.subplots_adjust(hspace=0.2)
plt.rc('font', size=32)

master_dir      = "***/Data"
colors          = ['#003876', '#00763e', '#760038', '#763e00']
state_label     = ['GS', '1st', '2nd', '3rd']
markers         = ['o', '^', 's']
dist            = np.around(np.arange(0.5, 2.01, 0.1), 3)

# Load
E_CAS       = np.zeros((len(dist), 3))
D_CAS       = np.zeros((len(dist), 3, 3))
C_qEOM      = np.zeros_like(E_CAS)
D_qEOM      = np.zeros_like(D_CAS)
S_qEOM      = np.zeros_like(E_CAS)
for inddist, dist in enumerate(dist):
    C_qEOM[inddist] = np.load(f"{master_dir}/H2/Hermitian_Contamination/C_{dist}.npy") / 2
    D_CAS[inddist]  = np.load(f"{master_dir}/H2/Dipole/{dist}_CAS.npy")[:3, :3, 2]
    D_qEOM[inddist] = np.load(f"{master_dir}/H2/Dipole/{dist}_CASCI_MATCH_qEOM.npy")[:,:,2]
    S_qEOM[inddist] = np.load(f"{master_dir}/H2/VAC/{dist}_CASCI_MATCH_qEOM.npy")[0]
D_err       = abs(abs(D_CAS) - abs(D_qEOM)) / 0.393430

# Energy
plt.subplot(211)
for index in range(1, 3):
    plt.plot(C_qEOM[:, index], abs(S_qEOM[:, index]), marker=markers[index], color='none', markersize=15, linewidth=0, markeredgecolor=colors[index], markeredgewidth=3)
plt.xticks([0.35, 0.4, 0.45, 0.5], fontsize=0, color='white')
plt.yticks([0, 0.2, 0.4, 0.6], ['0.00', '0.20', '0.40', '0.60'])
plt.ylabel('$|S_{0n}^O|$\n(Arb. Unit)')
plt.grid()

# Dipole
plt.subplot(212)
for Rex in range(3):
    for Lex in range(Rex + 1):
        plt.plot(0.5*(C_qEOM[:, Rex] + C_qEOM[:, Lex]), D_err[:, Lex, Rex], markerfacecolor='none', marker=markers[Lex], markersize=15, linewidth=0, markeredgecolor=colors[Rex], markeredgewidth=3)
plt.xticks([0.35, 0.4, 0.45, 0.5])
plt.xlabel('$0.5(C_{m}^H + C_{n}^H)$\n(Arb. Unit)')
plt.ylabel('$|\Delta\mu_Z|$\n(Debye)')
plt.grid()

for i in range(3):
    plt.plot([], [], markerfacecolor='none', marker=markers[i], markersize=15, linewidth=0, markeredgecolor='black', markeredgewidth=3, label=r'$\langle$' + f'${i}|$')
for i in range(3):
    plt.plot([], [], color=colors[i], linewidth=5, label=f'$|{i}$' + r'$\rangle$')
plt.legend(ncol=3, frameon=False, bbox_to_anchor=(0.87, -0.3))

fig.align_ylabels()
plt.savefig('***/Data/Figs/Fig2_vsC.png', bbox_inches='tight', dpi=1000)
```
