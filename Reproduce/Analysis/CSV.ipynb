## Master

```python
import numpy as np

master_dir  = '***'

dist_lst    = {
    'H2'        : np.around(np.arange(0.5, 2.01, 0.1), 2),
    'H2_Large'  : np.around(np.arange(0.5, 2.01, 0.1), 2),
    'H2_Huge'   : np.around(np.arange(0.5, 2.01, 0.1), 2),
    'LiH'       : np.around(np.arange(1.0, 2.01, 0.1), 2),
    'LiH_Large' : np.around(np.arange(1.0, 2.01, 0.1), 2),
    'HeH'       : np.around(np.arange(0.5, 2.01, 0.1), 2),
    'BeH2'      : np.around(np.arange(0.7, 1.61, 0.1), 2),
    'NH3'       : np.around(np.arange(0.5, 2.01, 0.1), 2),
    'H2O'       : np.around(np.arange(0.7, 1.61, 0.1), 2),
    'H2O_Small' : np.around(np.arange(0.7, 1.61, 0.1), 2),
    'HF'        : np.around(np.arange(0.5, 2.01, 0.1), 2),
}

N_states    = {
    'H2'        : 3,
    'H2_Large'  : 3,
    'H2_Huge'   : 3,
    'LiH'       : 4,
    'LiH_Large' : 4,
    'HeH'       : 3,
    'BeH2'      : 3,
    'NH3'       : 3,
    'H2O'       : 3,
    'H2O_Small' : 3,
    'HF'        : 3,
}
```

## Major

```python
# Energy, Dipole
systems = ['H2', 'H2_Large', 'H2_Huge', 'LiH', 'LiH_Large', 'HeH', 'BeH2', 'NH3', 'H2O', 'H2O_Small', 'HF']

# Energy
for system in systems:
    dists   = dist_lst[system]
    N_state = N_states[system]
    E_lst   = np.zeros((3, len(dists), N_state))
    for ind, dist in enumerate(dists):
        E_lst[0, ind]   = np.load(f"{master_dir}/Data/{system}/Energy/{dist}_CAS.npy")[:N_state]
        E_lst[1, ind]   = np.load(f"{master_dir}/Data/{system}/Energy/{dist}_CASCI_MATCH_qEOM.npy")[:N_state]
        E_lst[2, ind]   = np.load(f"{master_dir}/Data/{system}/Energy/{dist}_CASCI_MATCH_fqEOM.npy")[:N_state]
    np.savetxt(f"{master_dir}/Data/CSV/Energy/{system}_CASCI.csv", E_lst[0], delimiter=",")
    np.savetxt(f"{master_dir}/Data/CSV/Energy/{system}_qEOM.csv" , E_lst[1], delimiter=",")
    np.savetxt(f"{master_dir}/Data/CSV/Energy/{system}_AqEOM.csv", E_lst[2], delimiter=",")

# Dipole
for system in systems:
    IND     = 0
    dists   = dist_lst[system]
    N_state = N_states[system]
    D_lst   = np.zeros((3, len(dists), N_state, N_state, 3))
    d_lst   = np.zeros((3, len(dists), sum(range(N_state+1))))
    for ind, dist in enumerate(dists):
        D_lst[0, ind]   = np.load(f"{master_dir}/Data/{system}/Dipole/{dist}_CAS.npy")[:N_state, :N_state]
        D_lst[1, ind]   = np.load(f"{master_dir}/Data/{system}/Dipole/{dist}_CASCI_MATCH_qEOM.npy")[:N_state, :N_state]
        D_lst[2, ind]   = np.load(f"{master_dir}/Data/{system}/Dipole/{dist}_CASCI_MATCH_fqEOM.npy")[:N_state, :N_state]
    D_lst /= 0.393430
    for R in range(N_state):
        for L in range(R + 1):
            d_lst[:, :, IND]    = abs(D_lst[:, :, L, R, 2])
            IND                += 1
    np.savetxt(f"{master_dir}/Data/CSV/Dipole/{system}_CASCI.csv", d_lst[0], delimiter=",")
    np.savetxt(f"{master_dir}/Data/CSV/Dipole/{system}_qEOM.csv" , d_lst[1], delimiter=",")
    np.savetxt(f"{master_dir}/Data/CSV/Dipole/{system}_AqEOM.csv", d_lst[2], delimiter=",")
```
